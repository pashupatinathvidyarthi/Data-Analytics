```python
import seaborn as sns
tips=sns.load_dataset("tips")
tips
```

```python
fig=go.Figure()
fig.add_trace(go.Scatter(x=tips.total_bill,y=tips.tip,mode='markers'))
```

```python
fig=go.Figure()
fig.add_trace(go.Scatter(x=tips.total_bill,y=tips.tip,mode='lines'))
```

```python
fig=go.Figure()
fig.add_trace(go.Bar(x=[1,2,3,4,5],y=[3,8,5,2,6]))
```

```python
fig=go.Figure()
fig.add_trace(go.Histogram(x=tips['total_bill']))
```

```python
fig=go.Figure()
fig.add_trace(go.Scatter(x=tips.total_bill,y=tips.tip,mode='markers',marker_size=12*tips['size']))
```

```python
fig=go.Figure()
fig.add_trace(go.Scatter3d(x=tips.total_bill,y=tips.tip,mode='markers',marker_size=12*tips['size'],z=tips['size']))
```

```python
fig=go.Figure()
fig.add_trace(go.Scatter3d(x=[1,2,3,4,5],y=[3,8,5,2,6],mode='lines',z=[1,2,7,3,5]))
```

```python
#Using Plotly Express

x=[1,2,3,4]
y=[10,13,15,18]

fig=px.scatter(x,y)
fig.show()
```

```python
x=[1,2,3,4]
y=[10,13,15,18]

fig=px.line(x,y)
fig.show()
```

```python
x=['A','B','C']
y=[5,10,3]
fig=px.bar(x=x,y=y)
fig.show()
```

```python

x=[5,10,3,4,5,5,6,6,10,10,3,11,11,12]
fig=px.histogram(x)
fig.show()
```

```python
x=['A','B','C']
y=[5,10,3,25,12,33]
fig=px.box(x,y)
fig.show()
```
