```python
#Loops
#Loops allows you to repeatedly execute a block of code, while and for loop

#while
#repeatedly executes a block of code as long as condition is true.
```

```python
n=5
i=1
while i<n:
    print(i)
    i+=1
```

```python
count=5
while count>0:
    print(count)
    count-=1
else:
    print("This will be executed when the while loop is run successfully.")
```

```python
#for loops
#iterate over a sequence of elements
```

```python
a='pashupati'
for char in a:
    print(char)
```

```python
l=['3',4,5,"Pashupati",'5',6,'Nath']
for i in l:
    print(i)
```

```python
for i in l:
    if i=="Nath":
        break
    print(i)
else:
    print("This will execute when for loop ends")
```

```python
for i in l:
    if i=="Pashupati":
        continue
    print(i)
else:
    print("This will execute when for loop ends")
```

```python
#print a right triangle using while loop
row=1
while row<=4:
    col=1
    while col<=row:
        print("*",end="")
        col+=1
    print()
    row+=1
```
