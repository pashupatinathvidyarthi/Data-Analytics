```python
#Loops
#Loops allows you to repeatedly execute a block of code, while and for loop

#while
#repeatedly executes a block of code as long as condition is true.
```

```python
n=5
i=1
while i<n:
    print(i)
    i+=1
```

```python
count=5
while count>0:
    print(count)
    count-=1
else:
    print("This will be executed when the while loop is run successfully.")
```
