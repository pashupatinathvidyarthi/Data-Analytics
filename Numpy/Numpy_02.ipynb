```python
#Conditions on Array

arr=np.random.randint(1,10,(5,6)) #5 rows and 6 columns of random integers between 1 to 10 (10 is exclusive)
```

```python
arr
```

```python
arr.shape
```

```python
arr.ndim
```

```python
arr.size
```

```python
arr>3 #returns a boolean array of the same shape as arr, where each element is True if the corresponding element in arr is greater than 3, and False otherwise.
```

```python
arr[arr>3] #returns a 1D array containing only the elements of arr that are greater than 3. The resulting array will have a shape that depends on the number of elements in arr that satisfy the condition arr > 3.
```

```python
arr[arr<3] #returns a 1D array containing only the elements of arr that are less than 3. The resulting array will have a shape that depends on the number of elements in arr that satisfy the condition arr < 3.
```

```python
#To access specific rows and columns in the array
#To accesss element at specific row and column, we can use the indexing operator [].
arr[0] #returns the first row of the array arr. The resulting array will have a shape that depends on the number of columns in arr.
```

```python
arr[1] #returns the second row of the array arr. The resulting array will have a shape that depends on the number of columns in arr.
```

```python
arr[1][0] #returns the first element of the second row of the array arr. The resulting value will be a single integer, which is the element located at the first column of the second row in arr.
```

```python
arr[0:2] #returns the first two rows of the array arr. The resulting array will have a shape that depends on the number of columns in arr.
```

```python
arr
```

```python
arr[0:3,[0,2]] #returns a 2D array containing the first three rows and the first and third columns of arr. The resulting array will have a shape that depends on the number of rows and columns in arr that satisfy the specified indices.
```

```python
arr
```

```python
#Slicing on both rows and Column
arr[0:2,1:3] #returns a 2D array containing the first two rows and the second and third columns of arr. The resulting array will have a shape that depends on the number of rows and columns in arr that satisfy the specified indices.
```

```python
arr
```

```python
arr[0:2,[0,1]] #returns a 2D array containing the first two rows and the first and second columns of arr. The resulting array will have a shape that depends on the number of rows and columns in arr that satisfy the specified indices.
```

```python
arr
```

```python
arr[0,[1,2]] #returns a 1D array containing the elements from the first row of arr at the second and third columns. The resulting array will have a shape that depends on the number of columns in arr that satisfy the specified indices.
```

```python
arr
```

```python
arr[2:4,[0,3]] #returns a 2D array containing the third and fourth rows and the first and fourth columns of arr. The resulting array will have a shape that depends on the number of rows and columns in arr that satisfy the specified indices.
```

```python
arr[2:5] #returns the third, fourth and fifth rows of the array arr. The resulting array will have a shape that depends on the number of columns in arr.
```

```python
arr[2:5,0:2] #returns a 2D array containing the third, fourth and fifth rows and the first and second columns of arr. The resulting array will have a shape that depends on the number of rows and columns in arr that satisfy the specified indices.
```

```python
#Mathematical operations on arrays
arr1=np.random.randint(1,3,size=(3,3)) #3 rows and 3 columns of random integers between 1 to 3 (3 is exclusive)
arr2=np.random.randint(1,3,size=(3,3)) #3 rows and 3 columns of random integers between 1 to 3 (3 is exclusive)
```

```python
arr1
```

```python
arr2
```

```python
arr1+arr2 #returns a new array where each element is the sum of the corresponding elements from arr1 and arr2. The resulting array will have the same shape as arr1 and arr2, which is (3, 3) in this case. Each element in the resulting array will be the sum of the corresponding elements from arr1 and arr2.
```
