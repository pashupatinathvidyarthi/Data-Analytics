```python
#Expanding dimensions
arr=np.array([1,2,3,4])
```

```python
arr.ndim
```

```python
np.expand_dims(arr,axis=0) #Expand the shape of an array by inserting a new axis at the specified position. It returns a new array with the same data but with an additional dimension.
```

```python
np.expand_dims(arr,axis=1) #Expand the shape of an array by inserting a new axis at the specified position. It returns a new array with the same data but with an additional dimension.
```

```python
np.expand_dims(arr,axis=1).ndim
```
