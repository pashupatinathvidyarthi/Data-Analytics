```python
#access rows and column of a dataframe >> 
# implicit index/internal indexing/integer indexing
# explicit index/named indexing 
```

```python
df.iloc[0:2] # implicit index/internal indexing/integer indexing
#staring from 0 to 1, 2 is not included
```

```python
df
```

```python
df.loc[0:2] # explicit index/named indexing
#staring from 0 to 2, 2 is included
```

```python
df
```

```python
df.iloc[0:2]
```

```python
df.iloc[0:2,['Name','Parch']] #It will give an error because we are using implicit index and we are trying to access the column by name.
#Accessing column by name is only possible with explicit index/named indexing.
#It means iloc is only for accessing rows by integer index, it cannot be used to access columns by name.
```

```python
df.loc[0:2,['Name','Parch']] #It will give the desired output because we are using explicit index/named indexing and we are trying to access the column by name.
```

```python
df
```

```python
df.iloc[0:2,3:6] #It will give the desired output because we are using implicit index and we are trying to access the column by integer index.
```
