```python
x=[1,2,3,4,5]
y=[5,1,7,8,2]
plt.plot(x,y)
```

```python
plt.plot(x,y)
plt.title('Line Plot to understand the syntax')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
plt.plot(x,y, color='red') #To change the color of the line in the plot, we can use the color parameter in the plt.plot() function. By setting color='red', we specify that we want the line to be red. You can choose from a variety of colors such as 'blue', 'green', 'yellow', etc., or you can use hexadecimal color codes for more specific colors.
plt.title('Line Plot to understand the syntax')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
plt.plot(x,y, color='red',marker='o') #To change the color of the line in the plot, we can use the color parameter in the plt.plot() function. By setting color='red', we specify that we want the line to be red. You can choose from a variety of colors such as 'blue', 'green', 'yellow', etc., or you can use hexadecimal color codes for more specific colors. The marker parameter is used to specify the style of the markers that will be used to represent the data points on the plot. By setting marker='o', we specify that we want circular markers for each data point.
plt.title('Line Plot to understand the syntax')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
plt.plot(x,y, color='red',marker='o', linestyle='--')   #To change the color of the line in the plot, we can use the color parameter in the plt.plot() function. By setting color='red', we specify that we want the line to be red. You can choose from a variety of colors such as 'blue', 'green', 'yellow', etc., or you can use hexadecimal color codes for more specific colors. The marker parameter is used to specify the style of the markers that will be used to represent the data points on the plot. By setting marker='o', we specify that we want circular markers for each data point. The linestyle parameter is used to specify the style of the line connecting the data points. By setting linestyle='dashed', we specify that we want a dashed line.
plt.title('Line Plot to understand the syntax')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
plt.plot(x,y, color='red',marker='o', linestyle='--', linewidth=2)   #To change the color of the line in the plot, we can use the color parameter in the plt.plot() function. By setting color='red', we specify that we want the line to be red. You can choose from a variety of colors such as 'blue', 'green', 'yellow', etc., or you can use hexadecimal color codes for more specific colors. The marker parameter is used to specify the style of the markers that will be used to represent the data points on the plot. By setting marker='o', we specify that we want circular markers for each data point. The linestyle parameter is used to specify the style of the line connecting the data points. By setting linestyle='dashed', we specify that we want a dashed line. The linewidth parameter is used to set the width of the line. By setting linewidth=2, we specify that we want the line to be 2 points wide, which makes it thicker and more visible on the plot.
plt.title('Line Plot to understand the syntax')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
plt.plot(x,y, color='red',marker='o', linestyle='--', linewidth=2, markersize=7)   #To change the color of the line in the plot, we can use the color parameter in the plt.plot() function. By setting color='red', we specify that we want the line to be red. You can choose from a variety of colors such as 'blue', 'green', 'yellow', etc., or you can use hexadecimal color codes for more specific colors. The marker parameter is used to specify the style of the markers that will be used to represent the data points on the plot. By setting marker='o', we specify that we want circular markers for each data point. The linestyle parameter is used to specify the style of the line connecting the data points. By setting linestyle='dashed', we specify that we want a dashed line. The linewidth parameter is used to set the width of the line. By setting linewidth=2, we specify that we want the line to be 2 points wide, which makes it thicker and more visible on the plot. The markersize parameter is used to set the size of the markers. By setting markersize=10, we specify that we want the markers to be larger, making them more prominent on the plot.   
plt.title('Line Plot to understand the syntax')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
plt.plot(x,y, color='red',marker='o', linestyle='--', linewidth=2, markersize=7)   #To change the color of the line in the plot, we can use the color parameter in the plt.plot() function. By setting color='red', we specify that we want the line to be red. You can choose from a variety of colors such as 'blue', 'green', 'yellow', etc., or you can use hexadecimal color codes for more specific colors. The marker parameter is used to specify the style of the markers that will be used to represent the data points on the plot. By setting marker='o', we specify that we want circular markers for each data point. The linestyle parameter is used to specify the style of the line connecting the data points. By setting linestyle='dashed', we specify that we want a dashed line. The linewidth parameter is used to set the width of the line. By setting linewidth=2, we specify that we want the line to be 2 points wide, which makes it thicker and more visible on the plot. The markersize parameter is used to set the size of the markers. By setting markersize=10, we specify that we want the markers to be larger, making them more prominent on the plot.   
plt.title('Line Plot to understand the syntax')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid('True') #To add a grid to the plot, we can use the plt.grid() function. By setting grid=True, we specify that we want to display the grid on the plot. The grid helps to improve the readability of the plot by providing reference lines that make it easier to see the values of the data points and understand their positions relative to each other.
plt.show()
```

```python
x=[1,2,3,4,5]
y1=[5,1,7,8,2]
y2=[7,8,3,5,1]


plt.plot(x,y1, color='red',marker='o', linestyle='--', linewidth=2, markersize=7,label='y1')
plt.plot(x,y2, color='blue',marker='s', linestyle='--', linewidth=2, markersize=7,label='y2')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line Plot with Multiple Lines')
plt.legend() #To add a legend to the plot, we can use the plt.legend() function. The legend will display the labels for each line, which we specified using the label parameter in the plt.plot() function. This helps to differentiate between the different lines on the plot and provides clarity on what each line represents.
plt.grid('True')
plt.show()
```

```python
x=np.linspace(0,10,100) #To create an array of 100 evenly spaced values between 0 and 10, we can use the np.linspace() function from the NumPy library. The first argument (0) is the starting value, the second argument (10) is the ending value, and the third argument (100) specifies the number of values to generate. This will create an array of 100 values that are evenly spaced between 0 and 10, which can be used for plotting or other calculations.
```

```python
#plt.figure()  we take 1 blank page
plt.figure()
plt.subplot(1,2,1) #1 row, 2 columns, 1st plot
plt.plot(x,np.sin(x)) #To create a subplot, we can use the plt.subplot() function. The first argument (1) specifies the number of rows in the subplot, the second argument (2) specifies the number of columns, and the third argument (1) specifies that we want to plot on the first subplot. In this case, we are creating a 1x2 grid of subplots and plotting the sine function on the first subplot.

plt.subplot(1,2,2) #1 row, 2 columns, 2nd plot
plt.plot(x,np.cos(x)) #To create a subplot, we can use the plt.subplot() function. The first argument (1) specifies the number of rows in the subplot, the second argument (2) specifies the number of columns, and the third argument (2) specifies that we want to plot on the second subplot. In this case, we are creating a 1x2 grid of subplots and plotting the cosine function on the second subplot.
plt.show()
```
